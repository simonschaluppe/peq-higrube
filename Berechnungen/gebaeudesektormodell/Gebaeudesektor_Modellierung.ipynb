## Übergabe an die Analyse

- `data/scenarios.parquet`: gültige simulierte Typologie mit `batch` und `scenario_name`.
- `data/Building-Sector_Pool_1.xlsx`, Blatt `Strombed_estim`: je vollständigem Typenschlüssel eine Zeile; Spalten `Archetype`, `Nationale BGF heute`, `Nationale BGF 2040`. `Archetype` entspricht exakt `scenario_name` im Parquet. Bei mehrfach vorkommenden Szenarionamen zusätzlich `batch` angeben.
- Bestehende optionale Spalten `Bestand`, `Neubau`, `Sanierung`, `Energetischer Standard` können erhalten bleiben. Flächen in m² BGF. Nicht verwendete Typen erhalten ausdrücklich null; leere Fläche bedeutet ungeklärt.
- Blatt `zuordnung_gfz`, Spalten A:F: bestehende geometrische Zuordnung.

PV, Flexibilität, Wetter und Gebäudemasse müssen bereits in der vollständigen Typenzuordnung berücksichtigt sein. Die Analyse erzeugt dafür keine zusätzlichen Gewichte.
